## Configuration Parameters

Set the dataset, model, and attack parameters here.

```python
# Dataset selection: 0=Iris, 1=Crop, 2=Adult, 3=Breast Cancer, 4=Nursery, 5=Mushroom, 6=BC-TCGA
which_dataset = 0

which_model = 0 # Model selection: 0=Decision Tree, 1=Logistic Regression, 2=Multinomial NB, 3=KNN, 4=Random Forest, 5=MLP

# Explanation tool: 0=LIME, 1=SHAP
explanation_tool = 1

# Number of auxiliary datasets to experiment on
how_many_sets = 2  # Lower for faster runtime

# Size of auxiliary dataset (per class)
sample_set_sizes = [5]

# Query budget (total queries allowed)
query_limit = 100

# Number of experiments
num_experiments = 1

# Batch size for synthetic sample generation
batch_size = 10

# Save results option
save_option = False

print(f"Configuration:")
print(f"  Dataset: {which_dataset}")
print(f"  Model: {which_model}")
print(f"  Explanation Tool: {explanation_tool}")
print(f"  Query Limit: {query_limit}")
print(f"  Batch Size: {batch_size}")
```

## Run Generative Attack

Execute the generative attack using the configured parameters.

```python
# Run the attack
print("Starting generative attack...\n")

run_attack(
    wd=which_dataset,
    wm=which_model,
    et=explanation_tool,
    hms=how_many_sets,
    sss=sample_set_sizes,
    ql=query_limit,
    so=save_option,
    num_exp=num_experiments,
    bs=batch_size
)

print("\n Attack completed!")
```

## Experiment Variations

Run the attack with different configurations (datasets, models, batch sizes).

```python
# Run attack with multiple configurations
import numpy as np

explanation_tool = 1  # SHAP
how_many_sets = 2
sample_set_sizes = [5]
query_limit = 100
batch_size = 10

results_log = []

for which_dataset in [0, 2]:  # Test on Iris and Adult
    for which_model in [0, 1, 4]:  # Test on DT, LR, RF
        print(f"\n{'='*60}")
        print(f"Running attack: Dataset={which_dataset}, Model={which_model}")
        print(f"{'='*60}")
        
        try:
            run_attack(
                wd=which_dataset,
                wm=which_model,
                et=explanation_tool,
                hms=how_many_sets,
                sss=sample_set_sizes,
                ql=query_limit,
                so=save_option,
                num_exp=1,
                bs=batch_size
            )
            results_log.append({"dataset": which_dataset, "model": which_model, "status": "completed"})
        except Exception as e:
            print(f"\n❌ Error occurred: {e}")
            results_log.append({"dataset": which_dataset, "model": which_model, "status": f"failed: {str(e)}"})

print(f"\n{'='*60}")
print("Experiment Summary:")
print(f"{'='*60}")
for log in results_log:
    print(f"Dataset {log['dataset']}, Model {log['model']}: {log['status']}")
```

## Helper Functions

Utility functions for analysis and visualization.

```python
def get_dataset_name(dataset_idx):
    """Get human-readable dataset name."""
    names = {
        0: 'Iris',
        1: 'Crop',
        2: 'Adult Income',
        3: 'Breast Cancer',
        4: 'Nursery',
        5: 'Mushroom',
        6: 'BC-TCGA'
    }
    return names.get(dataset_idx, 'Unknown')

def get_model_name(model_idx):
    """Get human-readable model name."""
    names = {
        0: 'Decision Tree',
        1: 'Logistic Regression',
        2: 'Multinomial Naive Bayes',
        3: 'K-Nearest Neighbors',
        4: 'Random Forest',
        5: 'Multilayer Perceptron'
    }
    return names.get(model_idx, 'Unknown')

# Test helper functions
print(f"Dataset 0: {get_dataset_name(0)}")
print(f"Model 1: {get_model_name(1)}")
```
